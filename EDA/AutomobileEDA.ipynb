### Automobile EDA – Short Summary

I learned that customer car preferences vary across demographic and financial categories. Sedan was the most popular car category, followed by Hatchback and SUV. Male customers outnumbered female customers, and salaried customers were more common than business customers. I also discovered that factors such as profession, marital status, and loan status showed different patterns in car preferences. This analysis helped me understand how data cleaning, visualization, and univariate, bivariate, and multivariate analysis can reveal meaningful customer behavior and support data-driven business decisions.
